```python
#Eduardo Pérez
#Variables
x0, y0 = 15, 362.78
x1, y1 = 20, 517.35
x2, y2 = 22.5, 602.97
x_target = 21

#Newton coefficients
b0 = y0
b1 = (y1-y0)/(x1-x0)

f_x2_x1 = (y2-y1)/(x2-x1)
b2 = (f_x2_x1 - b1)/(x2-x0)

#Replacing in polynomial
v_21 = b0 + b1*(x_target-x0) + b2*(x_target-x0)*(x_target-x1)

print("La velocidad en 21 segundos es:",v_21,"m/s")
```

```python
#Eduardo Perez
import math

u, m0 = 2000, 140000
q, g = 2100, 9.8
a,b = 8,30

def f(t):
    return u*math.log(m0/(m0-q*t)) - g*t
def xTrue(t):
    mt = m0 - q*t
    logTerm = mt*math.log(m0/mt) + mt
    return -(u/q)*logTerm - 0.5*g*(t**2)

fA = f(a)
fB = f(b)
approxDistance = (b-a)*((fA+fB)/2)
trueDistance = xTrue(b)-xTrue(a)

absoluteError = trueDistance-approxDistance
relativeError = abs(absoluteError)/trueDistance

print(f"a) Distancia recorrida estimada: {approxDistance} m")
print(f"b) Absolute error: {absoluteError}")
print(f"c) Relative error: {relativeError}")
```

```python
#Eduardo Perez
import math

u, m0 = 2000, 140000
q, g = 2100, 9.8
a,b = 8,30
n = 2 

def f(t):
    return u*math.log(m0/(m0-q*t)) - g*t
def xTrue(t):
    mt = m0 - q*t
    logTerm = mt*math.log(m0/mt) + mt
    return -(u/q)*logTerm - 0.5*g*(t**2)

h = (b-a)/n

t0 = a
t1 = a+h
t2 = b

summation = 0
for i in range(1,n):
    summation += f(a+i*h)

coef = (b-a)/(2*n)
approxDistance = coef*(f(a)+2*summation+f(b))

trueDistance = xTrue(b)-xTrue(a)
absoluteError = trueDistance - approxDistance
relativeError = abs(absoluteError)/trueDistance

print("a) Distancia estimada: ",approxDistance)
print("b) Error absoluto: ", absoluteError)
print("c) Error relativo: ", relativeError)
```

```python
#Eduardo Perez
import math

def f(x):
    return 5*x*math.exp(-2*x)

def trueIntegral(x):
    return -(5/4)*math.exp(-2*x)*(2*x+1)

a,b = 0.1, 1.3

mid = (a+b)/2
h = (b-a)/2

fA, fMid, fB = f(a), f(mid), f(b)

approxVal = (h/3)*(fA + 4*fMid + fB)
trueVal = trueIntegral(b)-trueIntegral(a)

absoluteError = trueVal - approxVal
relativeError = abs(absoluteError)/trueVal

print("a) Integral por Simpsom: ",approxVal)
print("b) Error absoluto: ", absoluteError)
print("c) Error relativo: ", relativeError)
```

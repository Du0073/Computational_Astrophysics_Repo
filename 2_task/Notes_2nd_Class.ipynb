La función es:
$L(t) = L_0 e^{-t/\tau}$

Por lo tanto su derivada es:
$L'(t) = L_0 (-\frac{1}{\tau})e^{-t/\tau}$, es decir:

$L'(t) = -\frac{L_0}{\tau}e^{-t/\tau}$

Calculando el valor exacto de estas 2 funciones:

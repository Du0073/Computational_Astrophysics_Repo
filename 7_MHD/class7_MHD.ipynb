```python
import numpy as np
import matplotlib.pyplot as plt

# Simulation parameters for stable advection
NumPoints = 200
Length = 1.0
Velocity = 1.0
CFLNumber = 0.4
DeltaX = Length / NumPoints
DeltaT = CFLNumber * DeltaX / Velocity

# Grid initialization and initial wave condition
XCoords = np.linspace(0, Length, NumPoints, endpoint=False)
WaveField = np.sin(2 * np.pi * XCoords / Length)

# Time integration loop (Forward-Time Backward-Space scheme)
for _ in range(200):
    BackwardDiff = WaveField - np.roll(WaveField, 1)
    WaveField += -(Velocity * DeltaT / DeltaX) * BackwardDiff

# Visualization
plt.figure()
plt.plot(XCoords, WaveField, color='skyblue', label='CFL = 0.4')
plt.title("Original")
plt.xlabel("Posición (x)")
plt.ylabel("Amplitud (B)")
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulation parameters for unstable advection
NumPoints = 100
Length = 10.0
Velocity = 10.0
CFLNumber = 2.0  # CFL > 1 
DeltaX = Length / NumPoints
DeltaT = CFLNumber * DeltaX / Velocity

# Grid initialization and initial wave condition
XCoords = np.linspace(0, Length, NumPoints, endpoint=False)
WaveField = np.sin(2 * np.pi * XCoords / Length)

# Time integration loop (Forward-Time Backward-Space scheme)
for _ in range(200):
    BackwardDiff = WaveField - np.roll(WaveField, 1)
    WaveField += -(Velocity * DeltaT / DeltaX) * BackwardDiff

# Visualization
plt.figure()
plt.plot(XCoords, WaveField, color='salmon', label='CFL = 2.0')
plt.title("Inestable")
plt.xlabel("Posición (x)")
plt.ylabel("Amplitud (B)")
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Author signature as a plot watermark
plt.text(0.95, 0.05, transform=plt.gca().transAxes, 
         fontsize=10, color='gray', ha='right', va='bottom', alpha=0.8)
plt.show()
#Eduardo Pérez
```

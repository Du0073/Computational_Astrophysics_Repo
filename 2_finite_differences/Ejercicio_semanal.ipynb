### Deducción de las Fórmulas de Diferencias Finitas para la Segunda Derivada $f''(x)$

Para deducir las fórmulas numéricas, partimos de la expansión en series de Taylor alrededor de un punto $x$:

**1. Diferencia Finita Centrada:**
Evaluamos la serie de Taylor un paso hacia adelante $(x+h)$ y un paso hacia atrás $(x-h)$:
$$f(x+h) = f(x) + hf'(x) + \frac{h^2}{2}f''(x) + \frac{h^3}{6}f'''(x) + \mathcal{O}(h^4)$$
$$f(x-h) = f(x) - hf'(x) + \frac{h^2}{2}f''(x) - \frac{h^3}{6}f'''(x) + \mathcal{O}(h^4)$$

Sumando ambas ecuaciones, los términos con derivadas impares ($f'$ y $f'''$) se cancelan por simetría:
$$f(x+h) + f(x-h) = 2f(x) + h^2f''(x) + \mathcal{O}(h^4)$$

Despejando la segunda derivada $f''(x)$, obtenemos la aproximación centrada con un error de orden $\mathcal{O}(h^2)$:
$$f''(x) = \frac{f(x+h) - 2f(x) + f(x-h)}{h^2} + \mathcal{O}(h^2)$$

**2. Diferencia Finita Hacia Adelante:**
Evaluamos la serie de Taylor en $(x+h)$ y $(x+2h)$:
$$f(x+h) = f(x) + hf'(x) + \frac{h^2}{2}f''(x) + \frac{h^3}{6}f'''(x) + \mathcal{O}(h^4)$$
$$f(x+2h) = f(x) + 2hf'(x) + 2h^2f''(x) + \frac{8h^3}{6}f'''(x) + \mathcal{O}(h^4)$$

Multiplicamos la primera ecuación por 2 y se la restamos a la segunda ecuación para eliminar el término $f'(x)$:
$$f(x+2h) - 2f(x+h) = -f(x) + h^2f''(x) + h^3f'''(x) + \mathcal{O}(h^4)$$

Despejando $f''(x)$, obtenemos la aproximación hacia adelante con un error de orden $\mathcal{O}(h)$:
$$f''(x) = \frac{f(x+2h) - 2f(x+h) + f(x)}{h^2} + \mathcal{O}(h)$$

**3. Diferencia Finita Hacia Atrás:**
De forma análoga, evaluamos en $(x-h)$ y $(x-2h)$ y restamos $f(x-2h) - 2f(x-h)$ para aislar la segunda derivada:
$$f''(x) = \frac{f(x) - 2f(x-h) + f(x-2h)}{h^2} + \mathcal{O}(h)$$

```python
import numpy as np

t0 = 12
h = 1

def F(t):
    return 100 + 15 * np.sin(np.pi * t / 6) + 3 * np.sin(np.pi * t / 2)

def F_prime(t):
    return 15 * (np.pi / 6) * np.cos(np.pi * t / 6) + 3 * (np.pi / 2) * np.cos(np.pi * t / 2)

print("1) --------------")
for t_val in [10, 11, 12, 13, 14]:
    print(f"F({t_val}) = ", F(t_val))

print("2) --------------")
D_forward = (F(t0 + h) - F(t0)) / h
D_backward = (F(t0) - F(t0 - h)) / h
D_centered = (F(t0 + h) - F(t0 - h)) / (2 * h)

print('D_forward  = ', D_forward)
print('D_centered = ', D_centered)
print('D_backward = ', D_backward)

print("3) --------------")
exact_F_prime = F_prime(t0)
error_forward = abs((D_forward - exact_F_prime) / exact_F_prime) * 100
error_backward = abs((D_backward - exact_F_prime) / exact_F_prime) * 100
error_centered = abs((D_centered - exact_F_prime) / exact_F_prime) * 100

print("F'(12) exact       = ", exact_F_prime)
print('Forward error (%)  = ', error_forward)
print('Backward error (%) = ', error_backward)
print('Centered error (%) = ', error_centered)
#print("Presented by: Eduardo Perez")
```
